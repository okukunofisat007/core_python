```python
# 7. UNION() - It returns a new set containing elements from both sets

set1 = {1,2,3}
set2 = {3,4,5}
result = set1.union(set2)                #Union combined both sets and remove duplicate
print(result)

#OR using the pipe | sign
result = set1|set2
print = (result)
```

```python
#8. INTERSECTION() 

set1 = {1,2,3}
set2 = {3,4,5}
result = set1.intersection(set2)                #Union combined both sets and remove duplicate
print(result)

#OR using the pipe & hamper sign
result = set1&set2
print = (result)
```

```python
# 9. DIFFERENCES

set1 = {1,2,3}
set2 = {3,4,5}
result = set1.difference(set2)                #Union combined both sets and remove duplicate
print(result)

#OR using the pipe - minus sign
result = set1 - set2
print = (result)
```

```python
#10. Symmetric_difference() - ELEMENT IN EITHER SET BUT NOT BOTH

set1 = {1,2,3}
set2 = {3,4,5}
result = set1.symmetric_difference(set2)               #Union combined both sets and remove duplicate
print(result)

#OR using the ^ sign
result = set1 ^ set2
print = (result)
```

```python
#11. ISSUBSET() - Checks if one set is a subset of another

set1 = {1,2}
set2 = {1,2,3,4}
print(set1.issubset(set2))

#Alternatively, use <=:
print(set1 <= set2)
```

```python
#12. ISSUPERSET() - Checks if one set is superset of another.returns true if the first set contains all elements of thesecond set.

set1 = {1,2}
set2 = {1,2,3,4}
print(set1.issuperset(set2))

#Alternatively, use <=:
print(set1 >= set2)
```

```python
# 4. BITWISE Operators
a = 5
b = 3

print(a & b)          #Binary 001
print(a | b)          #Binary 111
print(a ^ b)          #Binary 110
print(~a)             #Binary -(101+1)
print(a << 1)         #Binary 1010
print(a >> 1)         #Binary 10











```
